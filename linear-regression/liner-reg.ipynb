#### Exercise 2: Linear Regression in 1D
Here, you generate synthetic data, plot it, fit a model, and evaluate performance using Mean Squared Error (MSE).

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_regression
from sklearn.linear_model import LinearRegression
```

```python
# 1. Generate dataset
X, y, _ = make_regression(n_samples=100,
                             n_features=1,
                             n_informative=1,
                             noise=10,
                             coef=True,
                             random_state=0,
                             bias=100.0)
```

```python
plt.scatter(X, y, color= "blue", label="Data points")
plt.title("X (1 Dimension) and y")
plt.xlabel("X")
plt.ylabel("y")
plt.show()
```

```python
model = LinearRegression()
model.fit(X,y)
```

```python
m = model.intercept_
b = model.coef_[0]
print(f"y = {m} * X + {b}")

y_line = m*X + b
```

```python

# Sort X for plotting
X_sorted = np.sort(X, axis=0)
y_line = model.predict(X_sorted)

plt.scatter(X, y, color='blue', label='Data')
plt.plot(X_sorted, y_line, color='red', label='Fitted Line')

plt.xlabel("X")
plt.ylabel("y")
plt.title("Linear Regression")
plt.legend()
plt.show()
```

```python
y_pred = model.predict(X)

def compute_mse(y_true, y_pred):
    return np.mean((y_true - y_pred)**2)

mse = compute_mse(y, y_pred)
print(f"MSE with noise=10: {mse}")
```

```python
# 5. Repeat with noise=50
X_50, y_50, _ = make_regression(n_samples=100, n_features=1, n_informative=1, noise=50, coef=True, random_state=0, bias=100.0)
model.fit(X_50, y_50)
y_pred_50 = model.predict(X_50)
print(f"MSE with noise=50: {compute_mse(y_50, y_pred_50)}")
```

#### Exercise 4: Forecast Diabetes Progression  (Multi variate Linear Regression)
Applying linear regression to a real-world medical dataset.

```python
from sklearn.datasets import load_diabetes
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
```

```python
import pandas as pd
diabetes = load_diabetes(as_frame=True)

df = pd.DataFrame(diabetes.frame)
df.head()
```

```python
df.describe()
df.shape
```

```python
model = LinearRegression()
model.fit(x_train, y_train)
```

```python
print(f"Coefficients: {model.coef_}")
print(f"Intercept: {model.intercept_}")
```

```python
y_train_pred = model.predict(x_train)
y_test_pred = model.predict(x_test)

mse_train = mean_squared_error(y_train, y_train_pred)
mse_test = mean_squared_error(y_test, y_test_pred)

print(f"Mean Squared Error (Train): {mse_train}")
print(f"Mean Squared Error (Test): {mse_test}")
```

#### Exercise 5: Gradient Descent

The goal of this exercise is to understand how the Linear Regression algorithm finds the optimal coefficients.

The goal is to fit a Linear Regression on a one dimensional features data without using Scikit-learn. Let's use the data set we generated for the exercise 2:
Part 1: Greedy Approach (Grid Search)

```python
X, y, coef = make_regression(n_samples=100,
                         n_features=1,
                         n_informative=1,
                         noise=10,
                         coef=True,
                         random_state=0,
                         bias=100.0)
```

```python
plt.scatter(X, y, color='blue', label='Data points')
plt.title("Scatter Plot of X and y")
plt.xlabel("X")
plt.ylabel("y")
plt.show()
```

```python
# Reshape X if necessary: X_flat = X.reshape(1,-1)[0]
X_flat = X.reshape(1, -1)[0] if X.ndim > 1 else X

def compute_mse_grid(coefs, X, y):
    a, b = coefs
    y_preds = a * X + b
    return np.mean((y_preds - y) ** 2)
```

```python
# Create grid of 640,000 points
aa, bb = np.mgrid[-200:200:0.5, -200:200:0.5]
grid = np.c_[aa.ravel(), bb.ravel()]

# Compute losses for all grid combinations
losses = [compute_mse_grid(point, X_flat, y) for point in grid]
losses_reshaped = np.array(losses).reshape(aa.shape)
```

```python
f, ax = plt.subplots(figsize=(8, 6))
contour = ax.contourf(
    aa,
    bb,
    losses_reshaped,
    100,
    cmap="RdBu",
    vmin=0,
                    vmax=160000)
ax_c = f.colorbar(contour)
ax_c.set_label("MSE")

ax.set(aspect="equal",
    xlim=(-200, 200),
    ylim=(-200, 200),
    xlabel="$a$",
    ylabel="$b$")
```

Part 2: Gradient Descent Optimization

```python
# Initialize parameters and hyper-parameters
a_current, b_current = 0.0, 0.0  # Starting point
learning_rate = 0.1
n_iterations = 100
n = len(X_flat)

history = []

for _ in range(n_iterations):
    y_preds = a_current * X_flat + b_current
    
    # Compute partial derivatives
    d_a = (2 / n) * np.sum((y_preds - y) * X_flat)
    d_b = (2 / n) * np.sum(y_preds - y)
    
    # Update parameters using gradient descent step
    a_current -= learning_rate * d_a
    b_current -= learning_rate * d_b
    
    # Save the current a and b
    history.append([a_current, b_current])

# Convert history list into a 2D NumPy array
history_array = np.array(history)
```

```python
# Setup the grid (from the previous grid search step)
aa, bb = np.mgrid[-200:200:.5, -200:200:.5]
losses_reshaped = np.array(losses).reshape(aa.shape)

# Create the contour plot
f, ax = plt.subplots(figsize=(8, 6))
contour = ax.contourf(aa,
                    bb,
                    losses_reshaped,
                    100,
                    cmap="RdBu",
                    vmin=0,
                    vmax=160000)
ax_c = f.colorbar(contour)
ax_c.set_label("MSE")

# Overlay the gradient descent path (a vs b values)
ax.plot(history_array[:, 0], history_array[:, 1], 
        color='cyan', 
        marker='o', 
        markersize=3, 
        linestyle='-', 
        label='Gradient Descent Path')

ax.legend()
ax.set(aspect="equal",
    xlim=(-200, 200),
    ylim=(-200, 200),
    xlabel="$a$",
    ylabel="$b$")

plt.show()
```
