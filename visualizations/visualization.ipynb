```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import plotly.express as px
import plotly.graph_objects as go
```

```python
df1 = pd.DataFrame({
    'name': ['christopher', 'marion', 'maria', 'mia',
             'clement', 'randy', 'remi'],
    'age': [70, 30, 22, 19, 45, 33, 20],
    'gender': ['M', 'F', 'F', 'F', 'M', 'M', 'M'],
    'state': ['california', 'dc', 'california', 'dc',
              'california', 'new york', 'porto'],
    'num_children': [2, 0, 0, 3, 8, 1, 4],
    'num_pets': [5, 1, 0, 5, 2, 2, 3]
})

df1
```

```python
df1.plot(
    x='name',
    y='age',
    kind="bar"
    
)

plt.title('Age per name')
plt.xlabel('name')
plt.legend()
plt.show()
```

#### Exercise 2 — Pandas Scatter Plot

```python
df = pd.DataFrame({
        'name':['christopher','marion','maria','mia','clement','randy','remi'],
        'age':[70,30,22,19,45,33,20],
        'gender':['M','F','F','F','M','M','M'],
        'state':['california','dc','california','dc','california','new york','porto'],
        'num_children':[4,2,1,0,3,1,0],
        'num_pets':[5,1,0,2,2,2,3]
        })

df.plot(
    x='age',
    y='num_children',
    kind='scatter',
    c='red'
)

plt.title('scatter plot num_children and age')
plt.xlabel('age')
plt.xlabel('num_children')
plt.show()
```

#### Exercise 4

```python
left_data = [5, 7, 11, 13, 17]
right_data = [0.1, 0.2, 0.4, 0.8, -1.6]
x_axis = [0.0, 1.0, 2.0, 3.0, 4.0]
```

```python
fig, ax1 = plt.subplots()


# This creates another y-axis sharing the same x-axis.
ax2 = ax1.twinx()

ax1.plot(
    x_axis,
    left_data,
    color='black'
)

ax2.plot(
    x_axis,
    right_data,
    color='red'
)

ax1.set_xlabel('X')
ax1.set_ylabel('Left Data')
ax2.set_ylabel('Right Data')

plt.title('Two Variables with Different Scales')

plt.show()
```

#### Exercise 5

```python
import numpy as np


# 2 rows
# ×
# 3 columns
# =
# 6 axes
fig, axes = plt.subplots(2,3)


# turns the 2D structure into something easier to loop over

# axes[0]
# axes[1]
# axes[2]
# axes[3]
# axes[4]
# axes[5]
axes = axes.flatten()

for i, ax in enumerate(axes, start=1):

    ax.text(
        0.5,
        0.5,
        f'(2,3,{i})',
        ha='center'
    )
    # 2. Sets the Y-axis limits from 0 to 1
    ax.set_ylim(0, 1)

    # 3. Sets the ticks to increment by 0.25 (0, 0.25, 0.5, 0.75, 1.0)
    ax.set_yticks(np.arange(0, 1.01, 0.25))


    ax.set_title(f'Title {i}')


# hspace=0.5 (Height Space): Adds vertical padding between rows of subplots. This stops the titles of the bottom plots
# from running into the X-axis labels of the top plots.wspace=0.5 (Width Space): Adds horizontal padding between columns 
# of subplots. This stops the Y-axis labels of the right plots from overlapping with the data of the left plots.
plt.subplots_adjust(
    hspace=0.5,
    wspace=0.5
)

plt.show()

```

#### Exercise 6

```python
returns = np.random.randn(50)
price = 100 + np.cumsum(returns)

dates = pd.date_range(start='2020-09-01', periods=50, freq='B')
df3 = pd.DataFrame(zip(dates, price),
                  columns=['Date','Company_A'])
```

```python
fig = px.line(
    df3,
    x='Date',
    y='Company_A',
    title='Company A Price'
)

fig.update_layout(
    xaxis_title='Date',
    yaxis_title='Price'
)

fig.show()
```

```python
fig = go.Figure()

fig.add_trace(
    go.Scatter(
        x=df3['Date'],
        y=df3['Company_A'],
        mode='lines',
        name='Company A'
    )
)

fig.update_layout(
    title='Company_1 stock price',
    xaxis_title='Date',
    yaxis_title='Company_A'
)

fig.show()
```

#### Exercise 7: Plotly Box plots

```python
y0 = np.random.randn(50)
y1 = np.random.randn(50) + 1 # shift mean
y2 = np.random.randn(50) + 2

plt.plot(y0)
plt.show()
```

```python

plt.plot(y1)
plt.show()
```

```python

plt.plot(y2)
plt.show()
```

```python
fig = go.Figure()

fig.add_trace(
    go.Box(
        y=y0,
        name='Distribution A'
    )
)

fig.add_trace(
    go.Box(
        y=y1,
        name='Distribution B'
    )
)

fig.add_trace(
    go.Box(
        y=y2,
        name='Distribution C'
    )
)

fig.update_layout(
    title='Comparison of Distributions'
)

fig.show()
```

1. Bar plot       → compare categories
2. Scatter plot   → investigate relationships
3. Line plot      → show change/trend
4. Subplots       → organize multiple visualizations
5. Twin axes      → compare variables with different scales
6. Box plot       → compare distributions
